```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot')
sns.set(rc={'figure.figsize':(10,6)})

df = pd.read_excel('Retail-Supply-Chain-Sales-Dataset.xlsx')

print("Shape of Dataset:", df.shape)

print("\nFirst 5 Rows")
display(df.head())

print("\nDataset Information")
df.info()

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nStatistical Summary")
display(df.describe())

df.drop_duplicates(inplace=True)

df.dropna(inplace=True)

date_col = None

for col in df.columns:
    if "date" in col.lower():
        date_col = col
        break

if date_col:
    df[date_col] = pd.to_datetime(df[date_col])

    df["Month"] = df[date_col].dt.month_name()
    df["Quarter"] = df[date_col].dt.quarter
    df["Year"] = df[date_col].dt.year

print(df.head())

print("Mean")
print(df.mean(numeric_only=True))

print("\nMedian")
print(df.median(numeric_only=True))

print("\nMode")
print(df.mode().iloc[0])

print("\nStandard Deviation")
print(df.std(numeric_only=True))

if "Month" in df.columns and "Sales" in df.columns:

    monthly = df.groupby("Month")["Sales"].sum()

    monthly = monthly.reindex([
        'January','February','March','April','May','June',
        'July','August','September','October','November','December'
    ])

    plt.figure(figsize=(12,5))
    monthly.plot(marker='o')
    plt.title("Monthly Sales Trend")
    plt.xlabel("Month")
    plt.ylabel("Sales")
    plt.grid(True)
    plt.show()

if "Quarter" in df.columns and "Sales" in df.columns:

    quarter = df.groupby("Quarter")["Sales"].sum()

    quarter.plot(kind='line',marker='o')
    plt.title("Quarterly Sales")
    plt.xlabel("Quarter")
    plt.ylabel("Sales")
    plt.show()

gender = None

for col in df.columns:
    if "gender" in col.lower():
        gender = col

if gender:
    sns.countplot(data=df,x=gender)
    plt.title("Gender Distribution")
    plt.show()

age = None

for col in df.columns:
    if "age" in col.lower():
        age = col

if age:
    sns.histplot(df[age],bins=10,kde=True)
    plt.title("Customer Age Distribution")
    plt.show()

product = None

for col in df.columns:
    if "product" in col.lower():
        product = col

if product and "Sales" in df.columns:

    top = df.groupby(product)["Sales"].sum().sort_values(ascending=False).head(10)

    top.plot(kind='bar')
    plt.title("Top 10 Best Selling Products")
    plt.ylabel("Sales Revenue")
    plt.xticks(rotation=45)
    plt.show()

category = None

for col in df.columns:
    if "category" in col.lower():
        category = col

if category and "Sales" in df.columns:

    cat = df.groupby(category)["Sales"].sum()

    cat.plot(kind='bar')
    plt.title("Sales Revenue by Category")
    plt.ylabel("Sales Revenue")
    plt.show()

plt.figure(figsize=(10,6))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()

if "Sales" in df.columns:

    sns.histplot(df["Sales"],bins=20,kde=True)
    plt.title("Sales Distribution")
    plt.show()

if "Sales" in df.columns:

    sns.boxplot(y=df["Sales"])
    plt.title("Sales Outlier Detection")
    plt.show()

if category and gender and "Sales" in df.columns:

    pivot = pd.pivot_table(df,
                           values="Sales",
                           index=category,
                           columns=gender,
                           aggfunc="sum")

    sns.heatmap(pivot,annot=True,cmap="YlGnBu")

    plt.title("Category vs Gender Sales")
    plt.show()

print("="*50)
print("Business Recommendations")
print("="*50)

print("""
1. Increase inventory for top-selling products.
2. Focus marketing campaigns on high-performing categories.
3. Provide offers during low-sales months.
4. Target customer groups with higher purchasing behavior.
5. Reduce stock for low-performing products.
""")
```
